# **1. Industry and Problem Definition**

## 1.1 Industry Context

Modern aerospace organizations operate large-scale sensor networks, including satellites, telescopes, and ground-based monitoring systems. These platforms generate continuous streams of heterogeneous data such as telemetry signals, structured sensor readings, and image-based observations. Human analysts cannot manually inspect every event in real time, especially under high-volume or anomaly-rich conditions.

As aerospace systems become more autonomous and data-intensive, there is increasing demand for AI-assisted decision-support tools that help analysts triage signals, detect anomalies, and interpret uncertain observations. These tools must operate reliably under noisy conditions and provide transparent reasoning to support human oversight.

# 1.2 Problem Definition

This project addresses the problem of AI-assisted anomaly triage in aerospace monitoring pipelines.

The goal is not to replace human operators, but to design a system that:

* classifies incoming signals from multiple sensor modalities
* evaluates confidence and robustness under uncertainty
* simulates rare or edge-case conditions
* assists human analysts in prioritizing investigation

Such a system functions as a decision-support assistant that reduces cognitive load while preserving human accountability.

# 1.3 Why AI is Appropriate

Aerospace sensor data is high-volume, noisy, and multi-modal. Traditional rule-based systems struggle with variability and unexpected signal patterns. AI methods are appropriate because they can:
* learn statistical structure in complex signals
* generalize across sensor noise and variation
* detect subtle anomalies
* model uncertainty
* assist human decision-making at scale

However, AI must be integrated cautiously. In safety-relevant industries, AI systems must emphasize transparency, interpretability, and bounded autonomy rather than full automation.

# 1.4 Constraints and Ethical Considerations

Aerospace monitoring systems operate in environments where incorrect decisions may have operational, financial, or safety consequences. Key constraints include:

* uncertainty in sensor reliability
* incomplete or biased training data
* risk of false positives causing alert fatigue
* risk of false negatives missing critical events
* overreliance on automated recommendations

Ethical considerations include:

* maintaining human oversight
* avoiding overconfidence in AI outputs
* ensuring auditability and accountability
* preventing misuse of monitoring technology
* designing systems that fail safely under uncertainty

This project therefore emphasizes human-in-the-loop decision support, conservative escalation policies, and transparent reasoning rather than autonomous control.

## 3.4 Data Flow and Orchestration Logic

At runtime, the system follows this workflow:

1.   **Ingest event** (tabular + optional image) and run validation checks.
2.   Run the **structured classifier** for tabular signals and the image classifier for image signals (when present).
3.   Aggregate outputs into an **evidence bundle** containing predicted labels and confidence scores.
4.   If the evidence indicates **low confidence or elevated risk**, trigger the **simulation probe** to test stability under plausible variation.
5.   Pass evidence (and robustness report when available) to the **triage agent**, which produces the final recommendation and audit trail.
6.   A human analyst reviews the recommendation and decides next actions.

A simple, defensible orchestration policy is:

* If confidence is **high** and robustness appears **stable**, allow “Normal” (or “Suspicious”) with minimal escalation.

* If confidence is **low**, signals conflict, robustness appears **unstable**, or data integrity warnings exist, escalate to “Suspicious” or “Critical” and recommend human review.

## 3.5 Assumptions

* Incoming events are broadly representative of the sensing domain, but may include noise, missingness, and drift.
* Labels used for supervised learning are imperfect and may reflect sampling bias.
* Generative samples reflect the training distribution and are therefore not guaranteed to represent real-world edge cases; they are used only as a robustness exploration tool.
* The system is evaluated as a prototype decision-support workflow rather than a production deployment.

## 3.6 Design Tradeoffs

* **Modularity vs. end-to-end optimization**: modular components are easier to audit, explain, and replace, but may not maximize raw predictive performance.

* **Conservatism vs. alert fatigue**: escalating uncertain cases improves safety but can increase false positives and analyst workload.

* **Simulation value vs. misuse risk**: generative modeling supports stress testing, but synthetic data can be misinterpreted if treated as ground truth; the system therefore constrains how simulation outputs are used.

* **Transparency vs. complexity**: bounded agent logic and audit trails improve accountability, but add implementation complexity compared to a single model.

## 3.7 Boundaries of Capability and Responsibility

This system is explicitly bounded as:

* **Decision support only**: it recommends triage categories and actions; it does not execute operational interventions.

* **Human accountable**: a human analyst remains responsible for final decisions and downstream actions.

* **No claims of real-time operational readiness**: the prototype demonstrates system design and integration rather than production-grade performance guarantees.

* **No security or surveillance claims**: the focus is aerospace monitoring and responsible signal triage in an industry-appropriate context.

## 4.1 Shared Event Representation

This system uses a unified ```SensorEvent``` structure to represent incoming aerospace signals. The shared schema allows all modules to operate on the same object and makes the pipeline modular. This mirrors real-world system design, where consistent data interfaces are essential for integration.

````python
# shared event schema
from dataclasses import dataclass
import random
import numpy as np

random.seed(42) #stabilize randomness
np.random.seed(42)

@dataclass
class SensorEvent:
    event_id: int
    tabular_features: np.ndarray
    image_present: bool
    data_quality_flag: bool
````

## 4.2 Structured Classification Module

This module represents the structured signal classification workflow inspired by the prior capstone project: **Applied Machine Learning — Exoplanet Signal Classification**.

It simulates a supervised classifier that outputs a label and confidence score. The purpose is to demonstrate how numerical sensor features contribute evidence to the integrated decision pipeline.

````python
# structured classifier module
def structured_classifier(event: SensorEvent):
    confidence = random.uniform(0.4, 0.95)
    label = "Suspicious" if confidence < 0.65 else "Normal"

    return {
        "module": "structured_classifier",
        "label": label,
        "confidence": round(confidence, 2)
    }
````

## 4.3 Image Perception Module

This module represents the image-based perception workflow inspired by the prior capstone project: **Deep Learning Systems — Galaxy Morphology CNN**.

It handles image signals as a distinct modality and produces classification evidence when image data is present.

````python
# image classifier module
def image_classifier(event: SensorEvent):
    if not event.image_present:
        return {"module": "image_classifier", "skipped": True}

    confidence = random.uniform(0.4, 0.95)
    label = "Suspicious" if confidence < 0.7 else "Normal"

    return {
        "module": "image_classifier",
        "label": label,
        "confidence": round(confidence, 2)
    }
````

## 4.4 Evidence Aggregation (Predictions + Confidence Fusion)

This module fuses outputs from the structured and image classifiers into a single, system-level evidence summary.  
It computes:

- **overall_confidence**: combined confidence across modalities (weighted, with graceful degradation if one modality is missing)
- **disagreement**: whether modalities conflict (label mismatch or large confidence gap)
- **risk_score**: an operational risk proxy used to decide whether to run the robustness probe

This corresponds to the **Evidence aggregation** block in Figure 1.


````python
def aggregate_evidence(structured_out: dict, image_out: dict, event: SensorEvent,
                       w_struct: float = 0.6, w_img: float = 0.4) -> dict:
    """
    Combine modality outputs into a single evidence summary for downstream routing.
    """
    rationale = []

    s_conf = structured_out.get("confidence", None)
    s_label = structured_out.get("label", None)

    img_skipped = image_out.get("skipped", False)
    i_conf = None if img_skipped else image_out.get("confidence", None)
    i_label = None if img_skipped else image_out.get("label", None)

    # --- Overall confidence (graceful degradation) ---
    if i_conf is None and s_conf is not None:
        overall_conf = float(s_conf)
        rationale.append("Image modality missing → overall confidence uses structured model only.")
    elif s_conf is None and i_conf is not None:
        overall_conf = float(i_conf)
        rationale.append("Structured modality missing → overall confidence uses image model only.")
    else:
        overall_conf = float(w_struct * s_conf + w_img * i_conf)
        rationale.append(f"Weighted fusion: {w_struct:.1f}*structured + {w_img:.1f}*image.")

    # --- Disagreement detection ---
    disagreement = False
    if (i_conf is not None) and (s_label is not None) and (i_label is not None):
        if s_label != i_label:
            disagreement = True
            rationale.append("Modalities disagree on label.")
        if abs(float(s_conf) - float(i_conf)) >= 0.25:
            disagreement = True
            rationale.append("Large confidence gap between modalities (>= 0.25).")
    elif i_conf is None:
        rationale.append("Disagreement check limited (image skipped).")

    # --- Risk score (operational proxy) ---
    # Base risk increases when confidence is low.
    risk = 1.0 - overall_conf

    # Disagreement increases operational risk.
    if disagreement:
        risk += 0.25

    # Data-quality flag is treated as a hard safety concern.
    if event.data_quality_flag:
        risk += 0.40
        rationale.append("Data quality flag raised → risk increased.")

    risk = float(min(max(risk, 0.0), 1.0))

    return {
        "module": "evidence_aggregation",
        "overall_confidence": overall_conf,
        "risk_score": risk,
        "disagreement": disagreement,
        "rationale": rationale,
        "signals": {
            "structured": {"label": s_label, "confidence": s_conf},
            "image": {"label": i_label, "confidence": i_conf, "skipped": img_skipped}
        }
    }
````

````python
def maybe_run_simulation_probe(event: SensorEvent, evidence: dict,
                               conf_threshold: float = 0.60,
                               risk_threshold: float = 0.70) -> dict:
    """
    Conditional routing to the robustness probe.
    Mirrors the 'Low confidence or high risk → Simulation probe' branch in Figure 1.
    """
    overall_conf = evidence["overall_confidence"]
    risk = evidence["risk_score"]
    disagreement = evidence["disagreement"]

    should_run = (overall_conf < conf_threshold) or (risk >= risk_threshold) or disagreement

    if not should_run:
        return {
            "module": "simulation_probe",
            "skipped": True,
            "reason": f"Skipped (conf={overall_conf:.2f}, risk={risk:.2f}, disagreement={disagreement}).",
            "stability": None
        }

    sim = simulation_probe(event)
    sim["skipped"] = False
    sim["reason"] = f"Triggered (conf={overall_conf:.2f}, risk={risk:.2f}, disagreement={disagreement})."
    return sim
````

## 4.5 Simulation and Robustness Probe

This module represents the generative robustness-testing ideas from the prior capstone project **Generative AI Applications — Synthetic Satellite Imagery with Variational Autoencoders**.

Instead of retraining a full generative model, the prototype simulates stability testing to demonstrate how generative reasoning influences system decisions.

### Deterministic Simulation Logic

The simulation probe uses deterministic behavior based on event ID to ensure reproducible evaluation scenarios. This allows controlled testing of how the system responds to stable versus unstable conditions without relying on randomness. The goal is to demonstrate architectural behavior rather than stochastic variation.

````python
# simulation probe (inspired by the VAE generative model project)
def simulation_probe(event: SensorEvent):
    # Deterministic stability based on event ID
    # Even IDs → stable
    # Odd IDs → unstable
    stability = "stable" if event.event_id % 2 == 0 else "unstable"

    return {
        "module": "simulation_probe",
        "stability": stability
    }
````

````python
def robustness_report(sim_out: dict) -> dict:
    """
    Convert simulation output into an explicit robustness artifact for the triage agent.
    """
    if sim_out.get("skipped", False):
        return {
            "module": "robustness_report",
            "skipped": True,
            "robust": None,
            "notes": [sim_out.get("reason", "Simulation skipped.")]
        }

    stable = sim_out.get("stability", "unknown")
    robust = True if stable == "stable" else False if stable == "unstable" else None

    notes = [
        "Robustness probe executed.",
        f"Stability outcome: {stable}."
    ]

    if robust is False:
        notes.append("Unstable under probe → treat as elevated operational risk.")

    return {
        "module": "robustness_report",
        "skipped": False,
        "robust": robust,
        "notes": notes
    }
````

### Note on Prototyping vs Full Model Re-Training

This capstone artifact is an **integration prototype**, not a full re-training of every prior model.
The goal is to demonstrate **system-level orchestration** and defensible decision logic grounded in earlier projects.

- The **structured** and **image** classifiers mirror the interfaces and outputs of prior trained models (label + confidence).
- The **simulation probe** represents the role of the VAE: stress-testing decisions under perturbation / counterfactual conditions.
- The **robustness report**, **audit log**, and **human-in-the-loop actions** are the *new integrative layer* that did not exist in any single prior project.

In a production version, the VAE probe would generate plausible variations of sensor imagery and/or telemetry, and stability would be computed from output consistency under these variations.


## 4.6 Agentic Decision Layer

This module implements the bounded triage logic inspired by the prior capstone project: **Agentic AI — Aerospace Anomaly Triage Workflow**.

It synthesizes classifier outputs and simulation results into a final recommendation, emphasizing safe escalation and transparent reasoning.

````python
def triage_agent(event: SensorEvent, evidence: dict, rob: dict) -> tuple:
    explanation = []

    # 1) Hard safety rule: data integrity issues always escalate
    if event.data_quality_flag:
        explanation.append("Data quality flag raised → immediate escalation for human review.")
        return "Critical", explanation

    risk = evidence["risk_score"]
    conf = evidence["overall_confidence"]
    disagreement = evidence["disagreement"]

    explanation.append(f"Evidence fusion: overall_confidence={conf:.2f}, risk_score={risk:.2f}.")
    if disagreement:
        explanation.append("Modalities disagree → treat as elevated uncertainty.")

    # 2) If robustness probe ran and found instability, escalate
    if (rob.get("skipped") is False) and (rob.get("robust") is False):
        explanation.append("Robustness probe indicates instability → escalate severity.")
        return "Critical", explanation

    # 3) Risk-based triage (bounded agentic rule set)
    if disagreement and conf < 0.80:
        explanation.append("Disagreement with non-high confidence → escalate to Suspicious.")
        return "Suspicious", explanation

    if risk >= 0.70 or conf < 0.55:
        explanation.append("High risk or low confidence → escalate to Suspicious.")
        return "Suspicious", explanation


    explanation.append("No integrity issue; confidence acceptable; probe stable or skipped → Normal.")
    return "Normal", explanation
````

````python
from datetime import datetime, timezone

def recommend_actions(decision: str, evidence: dict, rob: dict) -> list:
    """
    Human-in-the-loop action mapping (diagram: 'Recommended next actions').
    """
    actions = []

    if decision == "Normal":
        actions += [
            "Continue monitoring; no operator intervention required.",
            "Log event for trend analysis."
        ]
    elif decision == "Suspicious":
        actions += [
            "Request additional sensor capture (confirm signal persistence).",
            "Route to analyst queue for review within SLA window.",
            "Check recent similar events for clustering or recurrence."
        ]
        if evidence.get("disagreement", False):
            actions.append("Investigate modality disagreement (sensor alignment/calibration check).")
    elif decision == "Critical":
        actions += [
            "Escalate to on-call engineer immediately.",
            "Initiate incident protocol (isolate affected subsystem if applicable).",
            "Preserve raw telemetry + imagery for forensic analysis.",
            "Perform manual verification before any irreversible action."
        ]

    # If robustness probe flagged instability, add an explicit recommendation.
    if (rob.get("skipped") is False) and (rob.get("robust") is False):
        actions.append("Treat as high uncertainty: require human confirmation before decisions.")

    return actions


def build_audit_log(event: SensorEvent, structured_out: dict, image_out: dict,
                    evidence: dict, sim_out: dict, rob: dict,
                    decision: str, explanation: list,
                    thresholds: dict) -> dict:
    """
    Structured audit artifact (diagram: 'Explanation + audit log').
    """
    return {
        "timestamp_utc": datetime.now(timezone.utc).isoformat(),
        "event_id": event.event_id,
        "inputs": {
            "image_present": event.image_present,
            "data_quality_flag": event.data_quality_flag
        },
        "thresholds": thresholds,
        "module_outputs": {
            "structured_classifier": structured_out,
            "image_classifier": image_out,
            "evidence_aggregation": evidence,
            "simulation_probe": sim_out,
            "robustness_report": rob
        },
        "decision": decision,
        "explanation": explanation
    }
````

````python
def run_pipeline(event: SensorEvent,
                 conf_threshold: float = 0.60,
                 risk_threshold: float = 0.70) -> dict:
    # 1) Base classifiers
    s = structured_classifier(event)
    i = image_classifier(event)

    # 2) Evidence aggregation
    ev = aggregate_evidence(s, i, event)

    # 3) Conditional simulation probe
    sim = maybe_run_simulation_probe(
        event, ev,
        conf_threshold=conf_threshold,
        risk_threshold=risk_threshold
    )

    # 4) Robustness report
    rob = robustness_report(sim)

    # 5) Triage decision
    decision, explanation = triage_agent(event, ev, rob)

    # 6) Recommended actions + audit log
    actions = recommend_actions(decision, ev, rob)
    thresholds = {"conf_threshold": conf_threshold, "risk_threshold": risk_threshold}
    audit = build_audit_log(event, s, i, ev, sim, rob, decision, explanation, thresholds)

    return {
        "event_id": event.event_id,
        "decision": decision,
        "actions": actions,
        "explanation": explanation,
        "audit_log": audit
    }
````

## 5.7 Interpretation of Integrated System Behavior

The scenario suite confirms that the integrated triage architecture behaves consistently with its design objectives.

**Normal** outcomes occur when fused confidence is high and computed operational risk is low. In these cases, the system intentionally skips the robustness probe, reducing computational overhead and minimizing unnecessary analyst intervention. This demonstrates that conditional routing functions as a cost-aware orchestration mechanism rather than a static pipeline.

**Suspicious** outcomes arise when overall confidence is low or when multimodal signals disagree. In these cases, the robustness probe is triggered to evaluate stability under simulated perturbation. Even when stability is confirmed, the system escalates to human review, reflecting a conservative posture appropriate for safety-critical monitoring contexts.

**Critical** outcomes occur under two conditions:

(1) when data integrity flags activate a hard safety override independent of model confidence, and

(2) when instability is detected during the robustness probe. Notably, all unstable probe results lead to Critical classification, confirming that simulation-based robustness probing (inspired by the VAE project) meaningfully influences downstream triage decisions.

The risk–confidence visualization further supports this behavior, showing a clear inverse relationship between overall confidence and computed risk. Lower-confidence events cluster at higher risk levels and correspond to increased probe activation and, in cases of instability or integrity flags, escalation.

Overall, the results validate that multimodal evidence fusion, conditional robustness probing, and bounded agentic triage are integrated into a coherent, traceable decision process. The architecture behaves predictably under controlled scenarios, and each outcome remains auditable through structured logs.
